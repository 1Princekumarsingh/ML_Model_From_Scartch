```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
```

```python
class GradientBoostingRegressor:
  def __init__(self, n_estimators=100, learning_rate=0.01, max_depth=3):
    self.n_estimators = n_estimators
    self.learning_rate = learning_rate
    self.max_depth = max_depth
    self.models = [] #stores all trained trees
    self.init_prediction = None #stores the initial constant prediction

  def fit(self, X, y):
    self.init_prediction = np.mean(y) #starting point(reference prediction)
    y_curr_pred = np.full_like(y, self.init_prediction, dtype=float) #baseline

    for _ in range(self.n_estimators):
      residuals = y - y_curr_pred

      tree = DecisionTreeRegressor(max_depth = self.max_depth)
      tree.fit(X, residuals)

      update = tree.predict(X)
      y_curr_pred += self.learning_rate*update #Fm​(x)=Fm−1​(x)+η⋅hm​(x)

      self.models.append(tree)

  def predict(self, X):

    y_pred = np.full(X.shape[0], self.init_prediction, dtype=float) #same baseline

    for tree in self.models: # y^​(x)=F0​+m=1∑M​η⋅hm​(x)
      y_pred += self.learning_rate * tree.predict(X)

    return y_pred
```

```python
# Initial Prediction

# regressor : self.init_prediction = np.mean(y)

# classifier:
# p = np.mean(y)
# self.init_log_odds = np.log(p / (1 - p))
#       |
#       |
# prob = sigmoid(F)
# residuals = y - prob

# In classification, Gradient Boosting does NOT predict classes directly.
# Instead, it predicts log-odds.

# But humans (and loss functions) want probabilities, not log-odds.

# Sigmoid converts:
# log-odds → probability
# P(y=1) = σ(F(x))
```

```python
# Simple dataset
X = np.array([
    [1], [2], [3], [4], [5], [6]
])
y = np.array([1, 3, 2, 5, 4, 6])

model = GradientBoostingRegressor(
    n_estimators=50,
    learning_rate=0.1,
    max_depth=3
)

model.fit(X, y)
preds = model.predict(X)

print("Predictions:", preds)
```
